### Are Subsets views or copies in Pandas?

In pandas, whether you get a view (linked to original DataFrame) or a copy (independent) depends on how you slice or filter $\text{---}$ and it's not always obvious. 

- .iloc always returns a copy (safe, independent subset)
- .loc may return a view (especially if indexing on labels)

        subset = df.loc[df['Sex'] == 'male']
        This is like the earlier filtering - might be a view.

        so..
        subset = df.loc[df['Sex'] == 'male'].copy()
        
| Situation | View or Copy? | Should you use <span style="color: #F0D392;">.copy()?</span> |
| :--- | :--- | :--- |
| <span style="color: #F0D392;">.iloc[]</span> slicing | Mostly Copy | Optional |
| <span style="color: #F0D392;">.loc[]</span> slicing | May be View | Recommended |
| Filtering with condition <span style="color: #F0D392;">df[...]</span> | May be View | Recommended |
| You plan to modify the subset | Use <span style="color: #F0D392;">.copy()</span> | YES |

### Set Index v/s Re-Index

| Features | <span style="color: #F0D392;">set_index()</span> | <span style="color: #F0D392;">reindex()</span> |
| :--- | :--- | :--- |
| What it does? | **Moves a column** to become the index | **Changes the row/column labels** of a DataFrame |
| Main Use Case | Change the row labels using existing column | Match to new labels, add/remove/reorder rows/columns |
| Affects index | Yes $\text{--}$ creates a **new index from a column** | Yes $\text{--}$ but just **remaps or aligns to new index** |
| Adds NaNs? | No $\text{--}$ only restructures | Yes $\text{--}$ can insert NaNs for new labels |

#### 2. reindex() $\text{---}$ Conform to a new index/column structure

```python
sales_jan = pd.DataFrame({
    'Sales' : [250, 400, 300]
}, index=['Store_A', 'Store_B', 'Store_C'])

sales_jan
```

```python
sales_feb = pd.DataFrame({
    'Sales' : [260, 310]
}, index=['Store_A', 'Store_C'])

sales_feb
```

❌ Problem: Indexes are mismatched!

You cannot safely do operations like subtraction or direct comparison unless indexes are aligned.

✅ 🛠️ Solution: Use reindex() to align!

```python
common_index = sales_jan.index

print(common_index)
```

```python
sales_feb_aligned = sales_feb.reindex(common_index)

sales_feb_aligned
```

```python
sales_diff = sales_feb_aligned['Sales'] - sales_jan['Sales']

sales_diff
```

```python
sales_feb_aligned = sales_feb.reindex(common_index, fill_value=0)

sales_feb_aligned
```

```python
sales_diff = sales_feb_aligned['Sales'] - sales_jan['Sales']

sales_diff
```

### Iteration in Pandas
In pandas, iteration refers to looping over rows and columns of a DataFrame or Series. While pandas is optimized for vectorized operations, sometimes we need to loop $\text{---}$ for example, for conditional logic, debugging, or generating complex values.

```python
df = pd.DataFrame({
    'Name' : ['Alice', 'Bob', 'Charlie'],
    'Age' : [25, 30, 35],
    'City' : ['Delhi', 'Mumbai', 'Chennai']
})
```

```python
df
```

```python
for index, row in df.iterrows():
    print(f'{index} : {row['Name']} lives in {row['City']} and is {row['Age']} years old.')
```

```python
for row in df.itertuples():
    print(row)
    print(row.Index)
    print(f'{row.Name} lives in {row.City}, and is {row.Age} years old.', end='\n\n')
```

```python
# iterate over columns

for col in df.columns:
    print(f'Column : {col}')
    print(df[col].values, end='\n\n')
```

### Sorting in Pandas
Sorting helps you organize your data for better readability, analysis and logical operations $\text{---}$ whether by index or actual data values.

#### Why sorting matters?
You might want to:
- See top-performing products (e.g,, highest sales)
- Order records by name/date
- Prepare data before merging or visualizing

#### Two Main Types of Sorting
1. sort_values() $\rightarrow$ Sort by data values
2. sort_index() $\rightarrow$ Sort by index (row or column labels)

#### 1. sort_values() $\rightarrow$ Sort by data values

##### Syntax:

    DataFrame.sort_values(by, axis=0, ascending=True, inplace=False, na_position='last')

| Argument | Purpose |
| :--- | :--- |
| <span style="color: #F0D392;"> by </span> | column label(s) to sort by |
| <span style="color: #F0D392;"> axis </span> | 0 = rows (default), 1 = columns |
| <span style="color: #F0D392;"> ascending </span> | True = low to high, False = high to low |
| <span style="color: #F0D392;"> inplace </span> | True = modify current object |
| <span style="color: #F0D392;"> na_position </span> | 'first' or 'last' for NaNs |

```python
# Example of inplace

class Cal:
    def __init__(self, a):
        self.a = a

    def add(self, b):
        c = self.a + b
        return c

    def inplace_add(self, b):
        self.a += b
        return self.a
```

```python
c_cal = Cal(10)
```

```python
c_cal.a
```

```python
c_cal.add(10)
```

```python
c_cal.a
```

```python
c_cal.a = c_cal.add(10)     # inplace=False
```

```python
c_cal.a
```

```python
c_cal.inplace_add(10)       # inplace=True
```

```python
c_cal.a
```
