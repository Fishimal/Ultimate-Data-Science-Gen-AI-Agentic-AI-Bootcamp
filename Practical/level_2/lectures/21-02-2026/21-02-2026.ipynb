### Dot Product
$\sum\limits_{\text{i = 1}}^{\text{n}} a_i  \cdot b_i$

```python
print(a)
print(b)
```

```python
# with vectors
print(np.dot(a, b))
```

```python
# with matrices
a_2 = np.array([[1, 2], [3, 4]])
b_2 = np.array([[1, 2], [3, 4]])
```

```python
print(a_2)
print()
print(b_2)
```

$$
\begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix} 
\times 
\begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix} 
= 
\begin{bmatrix} (1 \times 1) + (2 \times 3) & (1 \times 2) + (2 \times 4) \\ (3 \times 1) + (4 \times 3) & (3 \times 2) + (4 \times 4) \end{bmatrix} 
= 
\begin{bmatrix} 7 & 10 \\ 15 & 22 \end{bmatrix}
$$

```python
print(np.dot(a_2, b_2))
```

#### Facts of Dot Product

**Fact 1:**

        - if matrix A is of shape (m, n)
        - and matrix B is of shape (n, p)
        - then np.dot(A, b) is of shape (m, p)

**Fact 2:**

        - number of columns in matrix A = number of rows in matrix B

**Fact 3:**

        - let's say if someone is trying to multiply 2 vectors in that case numpy automatically converts one vector into its transpose

```python
a_1 = np.array([[1, 2, 3], [4, 5, 6]])
b_1 = np.array([[1, 2, 3], [4, 5, 6]])
```

```python
print(a_1.shape)
print(b_1.shape)
```

```python
print(np.dot(a_1, b_1))
```

```python
print(np.dot(a_1, b_1.T))
```

```python
print(a_1 @ b_1.T)
```

### Sort, Search & Counting Functions

```python
arr = np.array([3, 6, 1, 1, 1, 1, 8, 2, 5, 6])
```

```python
print(arr)
```

```python
print(f'Sorted : {np.sort(arr)}')
print(f'Sorted Indices : {np.argsort(arr)}')
# Instead of returning the sorted value themseleves, it returns an array of indices that, when used to index the original array, would produce a sorted sequence.
```

```python
print(f'Insert 4 position : {np.searchsorted(np.sort(arr), 4)}')
print(f'Unique Elements : {np.unique(arr)}')
print(f'Count of Each Element : {np.bincount(arr)}')

# bincount() -> MIN_IDX = 0 & MAX_IDX = 8
# idx = [0 1 2 3 4 5 6 7 8]
# freq = [0 4 1 1 0 1 2 0 1]
```

```python
arr_new = np.array([3, 6, 1, 1, 1, 1, 1.2, 8, 2, 5, 6])
```

```python
print(arr_new)
```

```python
print(f'Count of Each Element : {np.bincount(arr_new)}')
```

### NumPy Char & Strings

#### np.char.<function_name>()

##### 1. np.char.add()



```python
a = np.array(['hello', 'world'])
b = np.array(['_user', '_123'])

print(np.char.add(a, b))
```

##### 2. np.char.multiply()

```python
print(np.char.multiply(a, 2))
```

##### 3. np.char.upper() / np.char.lower() / np.char.capitalize() / np.char.title()

```python
a = np.array(['hello', 'WORLD', 'hello WORLD'])

print(np.char.upper(a))
print(np.char.lower(a))
print(np.char.capitalize(a))
print(np.char.title(a))
```

##### 4. np.char.center()

```python
a = np.array(['hello', 'world'])
print(np.char.center(a, 11, fillchar='*'))
```

```python
def password(value):
    p = np.array([value[-4:]])
    print(np.char.center(p, 30, fillchar='*'))
```
