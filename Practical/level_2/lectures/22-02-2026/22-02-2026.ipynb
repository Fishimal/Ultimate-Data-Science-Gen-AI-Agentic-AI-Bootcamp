### What is Pandas?
Pandas is an open-source Python library for *data manipulation and analysis*. It gives you powerful, fast, flexible and expressive data structures designed to work with structured data.

It is built on top of NumPy, and its name comes from **"Panel Data"** *(econometrics term for multi-dimensional data)*.

### Why Pandas? What problem does it solve?

#### Before Pandas:

People used raw Python lists, dictionaries or NumPy arrays.

These structures lacked labels, indexing, grouping, join/merge, handling missing data, etc.

Real-world datasets (CSV, JSON, Excel, SQL) are structured in rows and columns $\text{---}$ not arrays.

So Pandas was designed to work like:

📊 Excel + 📚 NumPy + 🗂️ SQL + 🤖 ETL tools $\text{---}$ inside Python

### ⚙️ Where is Pandas used?

| Field | How Pandas Helps |
| :--- | :--- |
| Data Science | Load, clean, explore, transform and model data |
| Machine Learning | Prepare features, preprocesses, manipulate large datasets |
| Finance | Time series, stock prices, trade records |
| Web/Data Analytics | ETL pipelines, server logs, product usage stats |
| Healthcare | Patient records, diagnostics, hospital statistics |

### 🎁 And Tons of Utilities

- Reading/Writing (CSV, JSON, Excel, SQL, Parquet)
- Powerful slicing/indexing/merging
- GroupBy, pivoting, reshaping
- Time series and date handling
- Missing data handling
- Statistical functions and window operations
- Built-in plotting (with Matplotlib backend)

Think of Pandas like a spreadsheet that you can program, automate and scale $\text{---}$ with the power of Python and NumPy underneath 

### Series in Pandas

```python
import pandas as pd
```

#### 1. Creating Series From Lists

```python
# pd.Series(data, index=None, dtype=None, name=None, copy=False, fastpath=False)
# pd.Series(data)

data = [1, 2, 3, 4, 5, 6]
s1 = pd.Series(data)
```

```python
print(s1)
```

```python
s2 = pd.Series(data, index=['a', 'b', 'c', 'd', 'e', 'f'], name='Numbers')
```

```python
print(s2)
```

#### 2. Creating Series From Dictionaries

```python
data = {'a': 1, 'b': 2, 'c': 3}
s3 = pd.Series(data)
print(s3)
```

#### 3. Creating Series From NumPy Array

```python
import numpy as np
```

```python
arr = np.array([5, 10, 15])
s4 = pd.Series(arr, index=['x', 'y', 'z'])
print(s4)
```

#### 4. Series Attributes & Properties

```python
s5 = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'], name='Scores')
```

```python
print(s5)
print(s5.index)     # Index object
print(s5.values)    # NumPy array of values
print(s5.dtype)     # Data type
print(s5.name)      # Name of the Series
print(s5.shape)
print(type(s5.shape))
```

#### 5. Series Indexing & Access

##### 5.1. Positional Indexing

```python
# [10 20 30 40]
print(s5.values[0])
print(s5.values[3])
print(s5.values[4])
```

##### iloc
iloc gets rows (and/or columns) at integer locations.

```python
print(s5.iloc[0])
```

```python
print(s5.iloc[1:3])
```

##### 5.2. Label-based Indexing
loc gets rows (and/or columns) with particular labels.

```python
# Pandas resolves s2[0] in the following order:

# Check if 0 is a label in the index
# → If yes, return by label

# If not, and if the index is not integer-based, then treat 0 as a position
# → Return item at index position 0 (first element)

# FutureWarning: Series.__getitem__ treating keys as positions is deprecated. In a future version, integer keys will always be treated as labels (consistent with DataFrame behaviour). To access a value by position, use `ser.iloc[pos]`

print(s5[0])
```

```python
print(s5['a'])
```

```python
print(s5['b'])
```

```python
# .loc[] uses the label/index name, not position.
```

```python
print(s5[['b', 'd']])
```

```python
print(s5.loc[['b', 'd']])
```

```python
s6 = pd.Series([10, 20, 30, 40])
```

```python
s6.iloc[[1, 2]]
```

```python
s6.iloc[1:3]
```

```python
s6.loc[[1, 2]]
```

### DataFrame in Pandas

#### What is a DataFrame?
A DataFrame is a 2D labeled data structure with columns of potentially different types, just like a spreadsheet or SQL table.

Think of it like:
- A dict of Series
- Or a table with rows and columns

It is the core structure of pandas used for:
- Reading, cleaning, transforming, analyzing and exporting data

#### Why DataFrame?
Real-world data is usually tabular (e.g., CSV, Excel, SQL)

Rows = records

Columns = features

It supports label-based indexing, automatic alignment, missing data handling, grouping, aggregation and more
| Concept | Real-Life Analogy |
| :--- | :--- |
| <span style="color: #F0D392;"> DataFrame </span> | Excel sheet or database table |
| <span style="color: #F0D392;"> Series </span> | One column of that sheet |
| Index | Row number or ID |
| Columns | Named headers |

#### 1. Creating a DataFrame using A Dictionary of Lists

```python
data = {
    'Name' : ['Hemanth', 'Aditya', 'Rajbeer'],
    'Age' : [30, 31, 32],
    'City' : ['Delhi', 'Mumbai', 'Bengaluru']
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df
```

#### 2. Creating a DataFrame using A List of Dictionaries

```python
data = [
    {'Name' :  'Hemanth', 'Age' : 30, 'City' : 'Delhi'},
    {'Name' :  'Aditya', 'Age' : 31, 'City' : 'Mumbai'},
    {'Name' :  'Rajbeer', 'Age' : 32, 'City' : 'Bengaluru'}
]
```

```python
df = pd.DataFrame(data)
df
```

#### 3. Creating a DataFrame using NumPy Arrays (if all columns are of the same type)

```python
arr = np.array([[1, 2], [3, 4], [5, 6]])
df = pd.DataFrame(arr, columns=['A', 'B'])
df
```

#### 4. Accessing Data from DataFrame

```python
data = {
    'Name' : ['Hemanth', 'Aditya', 'Rajbeer'],
    'Age' : [30, 31, 32],
    'City' : ['Delhi', 'Mumbai', 'Bengaluru']
}

df = pd.DataFrame(data)
```

```python
print(df['Name'])       # Series (a column)
print(30 * '=')
print(df[['Name', 'Age']])      # DataFrame with 2 columns
print(30 * '=')
print(df.iloc[0])       # Row by integer index label or position
print(30 * '=')
print(df.loc[0])        # Row by index label
```

#### 5. Conditional Selection (Boolean Indexing)

```python
# Passengers older than 70
df['Age'] > 70      # boolean index of passengers > 70 (True) else False
```

```python
df[df['Age'] > 70]
```

```python
older_passengers = df[df['Age'] > 70]
display(older_passengers.head())
```

```python
older_passengers.shape
```

```python
older_passengers.info()
```

```python
older_passengers.describe()
```

```python
older_passengers['PassengerId'].describe()
```

```python
# Females who survived
df[(df['Survived'] == 1) & (df['Sex'] == 'female')]
```

```python
# get Name and Age Females who survived
female_survivors_subset = df[(df['Survived'] == 1) & (df['Sex'] == 'female')][['Name', 'Age']]
female_survivors_subset.head()
```

```python
# get Name and Age Females who survived using loc
# df.loc[labels, ['Name', 'Age']]

df.loc[(df['Survived'] == 1) & (df['Sex'] == 'female'), ['Name', 'Age']]
```
