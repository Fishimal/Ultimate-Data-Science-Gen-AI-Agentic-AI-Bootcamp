```python
a = list(range(1000000))
b = list(range(1000000))

start_p = time.time()
c = [x + y for x, y in zip(a, b)]
end_p = time.time()
total_p = end_p - start_p
print(f'Python: {total_p}')
```

```python
a = list(range(1000000))
b = list(range(1000000))

start_p = time.perf_counter()
c = [x + y for x, y in zip(a, b)]
end_p = time.perf_counter()
total_p = end_p - start_p
print(f'Python: {total_p}')
```

```python
import numpy as np
```

```python
a_np = np.array(a)
b_np = np.array(b)

start_n = time.perf_counter()
c_np = a_np + b_np
end_n = time.perf_counter()
total_n = end_n - start_n
print(f'NumPy: {total_n}')
```

```python
print(f'NumPy is faster than Python by : {(total_p / total_n):.2f}')
```

### Advantages of Numpy over Python

Python lists are:
- Flexible but slow for mathematical operations
- Not designed for matrix math
- Does not support vectorized operations
- Memory inefficient for large numerical data

Why use NumPy?
- Faster than pure Python for numerical computations
- Vectorized Operations (no explicit loops)
- Broadcasting (operate on arrays of different shapes)
- Memory efficient
- Comes with powerful linear algebra, random tools, etc.

Use of NumPy:
- Scientific python

### NumPy Array Attributes

#### 1. ndarray.shape

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
print(arr.shape)
```

#### 2. ndarray.ndim

```python
print(arr.ndim)
```

#### 3. ndarray.size
Returns the total number of elements in the array.

```python
print(arr.size)
```

#### 4. ndarray.dtype
Tells the data type of the array elements.

```python
print(arr.dtype)
```

#### 5. ndarray.itemsize
Returns the size in bytes for one element.

- 1 byte $\rightarrow$ 8 bits
- int8 $\rightarrow$ 1 * 8 bits
- int16 $\rightarrow$ 2 * 8 bits
- int32 $\rightarrow$ 4 * 8 bits
- int64 $\rightarrow$ 8 * 8 bits

```python
print(arr.itemsize)
```

#### 6. ndarray.nbytes
Total number of bytes consumed by the array.

```python
print(arr.nbytes)
```

#### 7. ndarray.T
Returns the transpose of the matrix.

```python
print(arr)      # 2 rows, 3 colums
```

```python
print(arr.T)      # 3 rows, 2 colums
```

```python
print(arr.T.shape)
```

### Array Reshaping & Resizing in NumPy

#### 1. reshape()
Change shape without changing data.

##### Notes:
- The total number of elements must remain the same
- reshape() returns a new view if possible (no copy)
- You can use -1 to let NumPy infer one dimension

```python
arr = np.array([1, 2, 3, 4, 5, 6])
print(arr.shape)
```

```python
# 2x3
# 3x2
# 6x1
```

```python
reshaped_1 = arr.reshape(2, 3)
print(reshaped_1)
```

```python
reshaped_2 = arr.reshape(3, 2)
print(reshaped_2)
```

```python
reshaped_3 = arr.reshape(1, 6)
print(reshaped_3)
print(reshaped_3.shape)
print(reshaped_3.ndim)
```

```python
reshaped_4 = arr.reshape(6, 1)
print(reshaped_4)
```

```python
array = np.array([1, 2, 3, 4, 5, 6, 56, 100, 789])
print(array)
print(array.shape)
```

```python
reshaped_5 = array.reshape(3, -1)
print(reshaped_5)
print(reshaped_5.shape)
```

```python
reshaped_6 = array.reshape(-1, 3)
print(reshaped_6)
print(reshaped_6.shape)
```

```python
reshaped_7 = array.reshape(2, -1)
print(reshaped_7)
print(reshaped_7.shape)
```

```python
reshaped_7 = array.reshape(-1, -1)
print(reshaped_7)
print(reshaped_7.shape)
```
