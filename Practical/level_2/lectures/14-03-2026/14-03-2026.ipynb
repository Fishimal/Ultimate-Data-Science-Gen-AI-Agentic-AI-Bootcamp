### Creating Your Own Database

```python
# Step 1:
conn = sqlite3.connect('student_1.db')
```

```python
# Step 2:
cursor = conn.cursor()
```

```python
# Step 3:
cursor.execute('PRAGMA foreign_keys = ON;')
```

```python
cursor.execute('''
    CREATE TABLE ultimate_data_science (
        student_id INT PRIMARY KEY NOT NULL,
        name TEXT NOT NULL,
        batch_id INT NOT NULL
    );
''')
```

```python
# insert values using tuples

student_vals = [(1, 'Agi Ann Maria', 2),
                (2, 'Tanmay', 2),
                (3, 'Satyajit Roy', 2)
                ]
```

```python
for student_val in student_vals:
    cursor.execute('''
        INSERT INTO ultimate_data_science (student_id, name, batch_id)
        VALUES (?, ?, ?)
    ''', student_val)
```

```python
# Save --> push the data from memory to the hard-drive
# saves the state of database
conn.commit()
```

```python
conn.close()
```

```python
cursor.execute('SELECT * FROM ultimate_data_science;')

rows = cursor.fetchall()
print(rows)
```

```python
# List all tables

cursor.execute("SELECT name FROM sqlite_schema WHERE type='table' AND name NOT LIKE 'sqlite_%';")
rows = cursor.fetchall()
print(rows)
```

```python
# View Schema

cursor.execute("SELECT sql FROM sqlite_schema WHERE name='ultimate_data_science';")
rows = cursor.fetchall()
print(rows)
```
