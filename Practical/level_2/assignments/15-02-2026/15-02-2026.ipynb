## NumPy: Real-World Assignment (15-02-2026-UDS-2.0)

```python
import numpy as np
```

### 1. A 1D array of numbers from 0 to 9

```python
arr_1_d = np.arange(0, 10)
print(arr_1_d)
```

### 2. A 3x3 matrix of all 5s

```python
matrix_3_x_3 = np.full((3, 3), 5)
print(matrix_3_x_3)
```

### 3. A 4x4 identity matrix

```python
identity_matrix_4_x_4 = np.eye(4, 4)
print(identity_matrix_4_x_4)
```

### 4. 10 evenly spaced numbers between 1 and 100

```python
linspace_1 = np.linspace(1, 100, 10)
print(linspace_1)
```

### 5. A 2x2 array with random values between 0 and 1

```python
random_matrix_2_x_2 = np.random.rand(2, 2)
print(random_matrix_2_x_2)
```
