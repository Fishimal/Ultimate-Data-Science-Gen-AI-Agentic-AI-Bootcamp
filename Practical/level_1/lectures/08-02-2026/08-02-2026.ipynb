```python
class TV:
    def start_tv(self):
        print('TV')


class Microwave:
    def start_mw(self):
        print('MW')


class Fan:
    def start(self):
        print('Fan')
```

```python
# polymorphism concept

class TV:
    def start(self):
        print('TV')


class Microwave:
    def start(self):
        print('MW')


class Fan:
    def start(self):
        print('Fan')
```

```python
electrical_appliances = [TV(), Microwave(), Fan()]
```

```python
for appliance in electrical_appliances:
    appliance.start()
```

```python
from abc import ABC, abstractmethod
```

```python
class ElectronicContract(ABC):

    @abstractmethod
    def start(self):
        pass
```

```python
class TV(ElectronicContract):
    def start(self):
        print('TV')


class Microwave(ElectronicContract):
    def start(self):
        print('MW')


class Fan(ElectronicContract):
    def start(self):
        print('Fan')


class AirFrier(ElectronicContract):
    def start_air_frier(self):
        print('Air Frier')
    
    def start(self):
        print('Air Frier')
```

```python
electrical_appliances = [TV(), Microwave(), Fan(), AirFrier()]
```

```python
for appliance in electrical_appliances:
    appliance.start()
```
