#### \_\_str\_\_() & \_\_repr\_\_()

```python
class Car:
    def __init__(self, manufacturer, car_name):
        self.manufacturer = manufacturer
        self.car_name = car_name
        print('CAR')

    def __str__(self):
        return f'Manufacturer Name : {self.manufacturer} and we produce cars.'

    def __repr__(self):     # official representation of an object
        return f'Class Car\nManufacturer Name : {self.manufacturer} and we produce cars. This object contains the car model : {self.car_name}'
```

```python
car_1 = Car('Toyota', 'Glanza')
```

```python
car_1
```

```python
str(car_1)      # string representation of the object
```

```python
repr(car_1)
```

```python
car_1
```

```python
print(f'{car_1!s}')     # __str__()
```

```python
print(f'{car_1!r}')     # __repr__()
```

```python
class Wallet:
    def __init__(self, cash):
        self.cash = cash

    def __repr__(self):
        return f'Class Wallet\nCash : {self.cash}'

    def __add__(self, other):
        return Wallet(self.cash + other.cash)

    def __sub__(self, other):
        return Wallet(self.cash - other.cash)

    def __mul__(self, other):
        return Wallet(self.cash * other.cash)
```

```python
w1 = Wallet(50)
w2 = Wallet(30)
combined_sum = w1 + w2      # combined_sum = w1.__add__(w2)
print(combined_sum.cash)
```

```python
w1 = Wallet(100)
w2 = Wallet(67)
combined_sum = w1.__add__(w2)
print(combined_sum.cash)
```

```python
# w1 = Wallet(50)
# w1 -> __init__()
# w1 -> self.cash
# w1 -> __add__()
```

#### \_\_getitem\_\_()

```python
class ATM:
    def __init__(self):
        self.transaction_history = []

    def do_transaction(self, value):
        self.transaction_history.append(value)
        print(f'Transaction done : {value}')

    def __len__(self):
        return len(self.transaction_history)

    def __getitem__(self, index):
        return self.transaction_history[index]
```

#### \_\_eq\_\_()

```python
class Wallet:
    def __init__(self, cash):
        self.cash = cash

    def __repr__(self):
        return f'Class Wallet\nCash : {self.cash}'

    def __add__(self, other):
        return Wallet(self.cash + other.cash)

    def __sub__(self, other):
        return Wallet(self.cash - other.cash)

    def __mul__(self, other):
        return Wallet(self.cash * other.cash)

    def __eq__(self, other):
        return (self.cash == other.cash)
```

```python
w1_new = Wallet(100)
w2_new = Wallet(100)
```

```python
w1_new == w2_new
```

```python
w1_new.__eq__(w2_new)
```

```python
w1_new.__eq__(Wallet(-100))
```

```python
# with property decorator -> newer version

class Account:
    def __init__(self, balance):
        self.__balance = balance

    # get the balance
    # getter
    @property
    def balance(self):
         print('Checking Balance')
         return self.__balance
    
    # set the balance
    # setter
    @balance.setter
    def balance(self, value):
        if (value <= 0):
            print('Security Alert! Bank Balance cannot be less than or equal to 0.')
        elif (value > 1000000):
            print('Security Alert! Large transactions need approval.')
        else:
            print('Changing Balance')
            self.__balance = value

    # delete the balance
    @balance.deleter
    def balance(self):
        print('ERROR : You cannot delete a bank balance.')
```

```python
class Account:
    def __init__(self, balance):
        self.__balance = balance

    # get the balance
    # getter
    def balance_get(self):
        print('Checking Balance')
        return self.__balance
    
    # set the balance
    # setter
    def balance_set(self, value):
        print('Changing Balance')
        self.__balance = value

    # delete the balance
    def balance_del(self):
        print('ERROR : You cannot delete a bank balance.')

    balance = property(fget=balance_get, fset=balance_set, fdel=balance_del)
```

```python
ac_new = Account(100)
ac_new.balance
```

```python
ac_new.balance = 500
```

```python
ac_new.balance
```

```python
ac_new.balance = 500
```

```python
ac_new.balance
```

#### enumerate()

```python
indexing = 0

for i in range(1, 21):
    if ((indexing % 2) == 0):
        print(i, end=' ')

    indexing += 1
```

### list & dict Comprehension

#### list Comprehension

```python
numbers = list(range(1, 21))
```

```python
numbers
```

```python
def even_numbers(l):
    final_even_list = []

    for num in l:
        if ((num % 2) == 0):
            final_even_list.append(num)

    return final_even_list
```

```python
even_n = even_numbers(numbers)
```

```python
even_n
```

```python
# list comprehension

[num for num in numbers if ((num % 2) == 0)]
```

```python
[num if ((num % 2) == 0) else -1 for num in numbers]
```
