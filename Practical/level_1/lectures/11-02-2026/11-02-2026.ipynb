### Returning a New Function from a Function (Decorator)

```python
def my_func(func):

    def wrap():
        print('START')
        func()
        print(func)
        print(id(func))
        print('END')

    return wrap     # memory location containing function object of wrap
```

```python
# add inside the argument below is having a different address than the add we are catching

add = my_func(add)
```

```python
print(f'{add}, {my_func}')
print(f'{id(add)}, {id(my_func)}')
```

```python
add()
```

```python
# here my_func --> is a decorator
```

#### Example - 1

```python
import time
```

```python
def my_decorator(func):

    def wrapper():
        t1 = time.time()
        func()
        t2 = time.time()

        total_time_taken = t2 - t1
        print(f'Function took : {total_time_taken:.2f} seconds')

    return wrapper
```

```python
def loop_and_multiply():
    total = 1
    for i in range(1, 11):
        total += (total * i)
        time.sleep(0.1)
    print(total)
```

```python
loop_and_multiply = my_decorator(loop_and_multiply)         # decorator with manual way
```

```python
loop_and_multiply()
```

```python
# decorator with syntactic sugar

# instead of writing

'''
    def loop_and_multiply():
        total = 1
        for i in range(1, 11):
            total += (total * i)
            time.sleep(0.1)
        print(total)


    def my_decorator(func):

        def wrapper():
            t1 = time.time()
            func()
            t2 = time.time()

            total_time_taken = t2 - t1
            print(f'Function took : {total_time_taken:.2f} seconds')

        return wrapper


    loop_and_multiply = my_decorator(loop_and_multiply)
'''


# python gives short-hand way

# Step-1: Create a decorator
def my_decorator(func):
    
    def wrapper():
        t1 = time.time()
        func()
        t2 = time.time()

        total_time_taken = t2 - t1
        print(f'Function took : {total_time_taken:.2f} seconds')

    return wrapper


# Step-3: Using short-hand way add @decorator_name on top of your base function
@my_decorator

# Step-2: Create a base function
def loop_and_multiply():
    total = 1
    for i in range(1, 11):
        total += (total * i)
        time.sleep(0.1)
    print(total)
```

```python
loop_and_multiply()
```

```python
@my_decorator
def add():
    time.sleep(3)
    print('Add')
```

```python
add()
```

#### Example - 2

```python
def decorator_excep(func):
    def wrapper():
        try:
            func()
        except Exception as e:
            print(e)
    return wrapper
```

```python
@decorator_excep
def add():
    a = input('Enter a number: ')
    print(a + 100)
```

```python
add()
```

#### Decorator Final Concept

```python
def decorator_excep_v2(func):
    def wrapper(*args, **kwargs):
        try:
            func(*args, **kwargs)
        except Exception as e:
            print(e)
    return wrapper
```

```python
@decorator_excep_v2
def add(*args, **kwargs):
    sum_result = sum(args) + sum(kwargs.values())
    print(f'Sum of {args} = {sum(args)}')
    print(f'Sum of {kwargs} = {sum(kwargs.values())}')
    print(sum_result)
```

```python
add(1, 2, 3, 4, 5, a=6, b=7, c=8, d=9, e=10)
```

```python
add(1, b=2, c=100)
```

```python
add(1, 2, 3, 4, 5, 6, b=2, c=100)
```

```python
add(1, 2)
```

```python
add(a=88.88)
```

```python
add()
```

```python
add('hi', 1)
```

```python
def decorator_excep_v3(func):
    def wrapper(*args, **kwargs):
        try:
            result = func(*args, **kwargs)
        except Exception as e:
            print(e)
        return result
    return wrapper
```

```python
@decorator_excep_v3
def add(*args, **kwargs):
    sum_result = sum(args) + sum(kwargs.values())
    print(f'Sum of {args} = {sum(args)}')
    print(f'Sum of {kwargs} = {sum(kwargs.values())}')
    return sum_result
```

```python
addition_result = add(1, 2, 3, 4, 5, 6, b=2, c=100)
```

```python
addition_result
```

### Static

*Static means it belongs to the class, not an object.*

- shared across all objects
- created once
- not tied to any instance
- **Accessibility:**
    - Can be accessed by class name directly
        - Standard method to access static: via class name
        - Optionally: via object

#### Static Variables

```python
class User:
    # static variable
    course_name = 'UDSB2'

    def __init__(self, batch_no):
        self.batch_number = batch_no
```

```python
obj_user_1 = User(1)
obj_user_2 = User(2)
```

```python
print(obj_user_1.batch_number)
print(obj_user_1.course_name)
```

#### Static Functions

Used for logical grouping

- no object required
- cleaner name calling

```python
class MathUtils:
    @staticmethod
    def add(a, b):
        return (a + b)

    @staticmethod
    def subtract_a_from_b(a, b):
        return (b - a)

    @staticmethod
    def subtract_b_from_a(a, b):
            return (a - b)
```

```python
MathUtils.add(1, 2)
```

```python
MathUtils.subtract_a_from_b(100, 2)
```

```python
MathUtils.subtract_b_from_a(100, 2)
```

```python
class MathUtils_New:
    @staticmethod
    def add(a, b):
        return (a + b)

    @staticmethod
    def subtract_a_from_b(a, b):
        return (b - a)

    @staticmethod
    def subtract_b_from_a(a, b):
            return (a - b)

    def class_method(self):
         print(self.add(1, 2))
         print(MathUtils_New.add(1, 2))
```

```python
MathUtils_New.add(1, 2)
```

```python
obj_maths_1 = MathUtils_New()
```

```python
obj_maths_1.class_method()
```
